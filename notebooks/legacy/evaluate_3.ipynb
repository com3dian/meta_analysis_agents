# Evaluate Three Extraction Methods on the Same Paper

Runs **Direct LLM**, **Static Workflow** (`label_then_direct`: labeller → tagged paper → same extraction prompt family as direct LLM), and **MAS** on the same paper (`study_id` = `Study#`),
all using the `wopke_100` intercropping standard with structured Pydantic output.

Includes number/table highlighting pre-processing and ROUGE-L evaluation against ground truth.

### PDF to Markdown setup (run once)

`data/wopke_100/paper_output/` is **gitignored** and must be populated locally.
Run the cell below once to convert the paper for the selected study.

````python
# Build mapping from annotation Study# to paper file paths
gt_df   = load_ground_truth()
mapping = build_study_paper_mapping(gt_df)
print(f'Mapped {len(mapping)}/100 studies to paper files')
````

````python
# Select which study to evaluate
study_id = 10  # <- Change this to test different papers

if study_id not in mapping:
    print(f'Study# {study_id} could not be mapped to a paper file!')
    print('Available study IDs:', sorted(mapping.keys()))
else:
    paper_path = mapping[study_id]
    paper_name = paper_path.split('/')[-1]
    print(f'Study# {study_id} -> {paper_name}')

    # Convert PDF to markdown if needed
    if paper_path.endswith('.pdf'):
        from src.experimentutils import convert_pdf_to_markdown
        paper_path = convert_pdf_to_markdown(paper_path)
        print(f'Converted to: {paper_path.split("/")[-1]}')
````

````python
# Standard and schema
standard  = METADATA_STANDARDS['wopke_100']
n_fields  = len(SchemaFactory()._parse_schema_string(standard))
n         = study_id  # alias for filename convention

provider_label = re.sub(r'[^A-Za-z0-9]+', '-', (LLM_PROVIDER or 'unknown').strip()).strip('-')
model_label = re.sub(r'[^A-Za-z0-9]+', '-', get_model_name().strip()).strip('-')
file_tag = f'{provider_label}_{model_label}_{n_fields}fields'

print(f'Standard : wopke_100 ({n_fields} fields)')
print(f'LLM      : {provider_label}/{model_label}')

# Ground truth for this study
gt_paper = gt_df[gt_df['Study#'] == study_id].reset_index(drop=True)
print(f'GT rows  : {len(gt_paper)}')

# Read and highlight paper text
raw_text      = read_paper_text(paper_path)
highlighted   = highlight_numbers_and_tables(raw_text)
context       = create_context(source=paper_path, name=f'study_{study_id}')

print(f'\nDocument length : {len(raw_text):,} chars')
print(f'Highlighted len : {len(highlighted):,} chars (numbers and tables marked)')
````

````python
# Shared Pydantic output schema
try:
    OutputSchema = SchemaFactory().create_from_standard(
        standard,
        record_class_name='WopkeRecord',
        output_class_name='WopkeOutput',
        records_key='yield_records',
    )
    print(f'OutputSchema : {OutputSchema.__name__}')
    record_cls = OutputSchema.model_fields['yield_records'].annotation.__args__[0]
    print(f'Record fields: {list(record_cls.model_fields.keys())}')
except Exception as e:
    OutputSchema = None
    print(f'Could not build OutputSchema ({e})')
````

````python
# MAS objective for wopke_100 intercropping standard
mas_objective = f'''You are an expert agricultural meta-analysis specialist.
Your task is to extract **intercropping experiment records** from a scientific research paper.
This meta-analysis compares crop yield under intercropping settings versus sole cropping.

**META-ANALYTIC SCHEMA (CRITICAL)**:
{standard}

**SCHEMA RULES**
- Use JSON keys as the EXACT field names in every record.
- Schema descriptions are guidance only — values must be concrete text extracted from the paper.
- Do not rename, add, or remove any schema fields.

**MULTI-RECORD RULE (CRITICAL)**
Each unique combination of crop pair × site × year × treatment level = one SEPARATE record.
Examples:
- 2 density levels × 2 N-treatments = 4 records for the same crop pair.
- Each row in a yield results table is typically a separate record.
Do NOT collapse table rows or merge treatment combinations into a single record.

**YIELD FIELD MAPPING (CRITICAL)**
- `unified yield sc 1` = sole-crop yield of Crop species 1
- `unified yield sc 2` = sole-crop yield of Crop species 2
- `unified yield ic 1` = intercropped yield of Crop species 1
- `unified yield ic 2` = intercropped yield of Crop species 2
Use explicit table headers, row labels, and footnotes to assign yields to the correct species.
If species assignment is ambiguous, set the ambiguous field to null and note the source in `Data source`.
Preserve numeric values exactly — do not round or average.

**SHARED FIELDS**
Fields that are constant across treatments (Year, Lat, Lon, Experimental design, species names,
Intercropping pattern, uniform nutrient inputs) must be filled identically in every record.
Set to null only if genuinely absent from the paper.

**EXAMPLE RECORD** (Barley/Pea paper — shows expected detail level):
```json
{{
  "Year of data": "1980",
  "Experimental design": "Split-plot",
  "Lat": "55.68", "Lon": "12.08",
  "Crop species 1": "Barley", "Crop species 2": "Pea",
  "Crop type 1": "Cereal", "Crop type 2": "Legume",
  "Intercropping pattern": "Mixed",
  "Density ic 1": "154", "Density ic 2": "30",
  "Density sc 1": "275", "Density sc 2": "63",
  "N input SC1": "0", "N input SC2": "0", "N input IC1": "0", "N input IC2": "0",
  "N total in IC": "0", "N Unit": "kg N/ha",
  "Data source": "Table 3,4",
  "unified yield sc 1": "4.12", "unified yield sc 2": "4.56",
  "unified yield ic 1": "3.4009", "unified yield ic 2": "1.3891",
  "Yield unit": "g/m2"
}}
```

**OUTPUT**: One record per unique treatment combination using exact schema field names.
Omit records only if the paper genuinely contains no supporting data.'''
````

## Evaluation Against Ground Truth (ROUGE-L)

````python
import numpy as np
import matplotlib.pyplot as plt

from src.experimentutils import (
    field_similarity_score,
    find_crop_swap_pairs,
    match_records_greedy,
    evaluate_method_scores,
    print_matching_pairs,
)
from src.experimentutils.eval_utils import _is_missing

print('Evaluation utilities loaded.')
````

````python
wopke_field_names = list(SchemaFactory()._parse_schema_string(standard).keys())
gt_cols_set   = set(gt_paper.columns)
shared_fields = [f for f in wopke_field_names if f in gt_cols_set]
print(f'Shared evaluation fields: {len(shared_fields)} / {len(wopke_field_names)}')
````

````python
method_paths = {
    'direct_llm':      path_direct,
    'static_workflow': path_workflow,
    'mas':             path_mas,
}

method_scores = {}
method_field_scores = {}
method_overall_scores = {}
method_match_counts = {}
mas_artifact_scores = {}

# Use GT record count as denominator to penalize missing extracted records.
total_records = len(gt_paper)

for method, csv_path in method_paths.items():
    if csv_path is None or not os.path.exists(str(csv_path)):
        print(f'[{method}] no output CSV -- skipping.')
        continue

    ext_df = pd.read_csv(csv_path)
    if ext_df.empty:
        print(f'[{method}] empty CSV -- skipping.')
        continue

    eval_cols = [c for c in shared_fields if c in ext_df.columns]
    if not eval_cols:
        print(f'[{method}] no shared columns -- skipping.')
        continue

    scores_df, overall_norm, per_field_norm, n_matches = evaluate_method_scores(
        ext_df=ext_df,
        gt_df=gt_paper,
        shared_cols=eval_cols,
        total_records_for_denominator=total_records,
    )

    method_scores[method] = scores_df
    method_field_scores[method] = per_field_norm
    method_overall_scores[method] = overall_norm
    method_match_counts[method] = n_matches

    print(
        f'[{method}] extracted={len(ext_df)}, matched={n_matches}, '
        f'reference={total_records}, fields={len(eval_cols)} '
        f'-> normalized ROUGE-L = {overall_norm:.3f}'
    )

# Optional diagnostic: score MAS step-1 artifact (`field_value_pairs`) as value recall
# against all non-missing GT values (normalized exact string overlap).
def _norm(v):
    return str(v).strip().lower()

try:
    import json
    import re
    import ast

    raw_pairs = workspace.get('field_value_pairs', []) if 'workspace' in globals() else []

    def _parse_pairs_artifact(x):
        if isinstance(x, list):
            return x
        if not isinstance(x, str):
            return []

        s = x.strip()

        # Remove wrapping quotes if artifact is a quoted blob
        if len(s) >= 2 and ((s[0] == s[-1] == "'") or (s[0] == s[-1] == '"')):
            s = s[1:-1].strip()

        # Strip markdown fences
        s = re.sub(r'^```(?:json)?\s*', '', s, flags=re.IGNORECASE)
        s = re.sub(r'\s*```$', '', s)

        # Try full-string JSON first
        try:
            return json.loads(s)
        except Exception:
            pass

        # Try extracting the first array payload
        m = re.search(r'\[.*\]', s, flags=re.DOTALL)
        if m:
            payload = m.group(0)
            try:
                return json.loads(payload)
            except Exception:
                try:
                    return ast.literal_eval(payload)
                except Exception:
                    return []

        return []

    raw_pairs = _parse_pairs_artifact(raw_pairs)
    print(f"[mas_step1_value_recall] parsed field_value_pairs count = {len(raw_pairs)}")

    step1_values = set()
    for p in raw_pairs or []:
        if isinstance(p, (list, tuple)) and len(p) >= 2:
            val = p[1]
            if not _is_missing(val):
                step1_values.add(_norm(val))

    gt_values = set()
    for col in shared_fields:
        if col in gt_paper.columns:
            for v in gt_paper[col].tolist():
                if not _is_missing(v):
                    gt_values.add(_norm(v))

    if gt_values:
        mas_artifact_scores['mas_step1_value_recall'] = len(step1_values & gt_values) / len(gt_values)
        print(
            f"[mas_step1_value_recall] matched_values={len(step1_values & gt_values)} / "
            f"gt_values={len(gt_values)} -> recall={mas_artifact_scores['mas_step1_value_recall']:.3f}"
        )

        # Value-level ROUGE-L diagnostic for Step 1 artifact.
        # For each GT value, take the best ROUGE-L F1 against all Step-1 extracted values,
        # then average over all GT values.
        from rouge_score import rouge_scorer

        scorer = rouge_scorer.RougeScorer(['rougeL'], use_stemmer=False)
        step1_list = list(step1_values)
        gt_list = list(gt_values)

        if step1_list:
            per_gt_best = []
            for gt_val in gt_list:
                best = 0.0
                for ext_val in step1_list:
                    score = scorer.score(gt_val, ext_val)['rougeL'].fmeasure
                    if score > best:
                        best = score
                per_gt_best.append(best)

            mas_artifact_scores['mas_step1_value_rougeL'] = float(sum(per_gt_best) / len(per_gt_best))
            print(
                f"[mas_step1_value_rougeL] avg_best_over_gt={mas_artifact_scores['mas_step1_value_rougeL']:.3f} "
                f"(gt_values={len(gt_list)}, step1_values={len(step1_list)})"
            )
        else:
            mas_artifact_scores['mas_step1_value_rougeL'] = 0.0
            print('[mas_step1_value_rougeL] no Step-1 values parsed (set to 0.0).')
    else:
        mas_artifact_scores['mas_step1_value_recall'] = float('nan')
        mas_artifact_scores['mas_step1_value_rougeL'] = float('nan')
        print('[mas_step1_value_recall] GT has no comparable values.')
except Exception as e:
    mas_artifact_scores['mas_step1_value_recall'] = float('nan')
    print(f'[mas_step1_value_recall] skipped due to error: {e}')

print('\nEvaluation complete.')
````

````python
# Inspect matching pairs for each method.
# Change show_fields to focus on specific columns, or set max_pairs to limit output.
# key_fields = [
#     'Year of data', 'Crop species 1', 'Crop species 2',
#     'Density ic 1', 'Density ic 2', 'Density sc 1', 'Density sc 2',
#     'unified yield sc 1', 'unified yield sc 2',
#     'unified yield ic 1', 'unified yield ic 2', 'Yield unit',
# ]
# display_fields = [f for f in key_fields if f in shared_fields]

display_fields = list(shared_fields)

for method, csv_path in method_paths.items():
    print(f'{"="*70}')
    print(f'  METHOD: {method}')
    print(f'{"="*70}')
    print_matching_pairs(
        csv_path=csv_path,
        gt_df=gt_paper,
        shared_cols=shared_fields,
        show_fields=display_fields,
    )
    print()
````

````python
if not method_overall_scores:
    print('No scores to plot.')
else:
    COLORS = {
        'direct_llm': '#4C78A8',
        'static_workflow': '#F58518',
        'mas': '#54A24B',
        'mas_step1_value_recall': '#B279A2',
        'mas_step1_value_rougeL': '#E45756',
    }
    all_fields = sorted({c.replace('rougeL_', '') for d in method_field_scores.values() for c in d.keys()})

    nf, nm = len(all_fields), len(method_overall_scores)
    bw, x = 0.8 / max(nm, 1), np.arange(nf)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(max(14, nf * 0.55), 10))

    for i, meth in enumerate(method_overall_scores.keys()):
        means = [method_field_scores[meth].get(f'rougeL_{f}', float('nan')) for f in all_fields]
        ax1.bar(x + i * bw, means, width=bw, label=meth, color=COLORS.get(meth, 'steelblue'), alpha=0.85)

    ax1.set_xticks(x + bw * (nm - 1) / 2)
    ax1.set_xticklabels(all_fields, rotation=45, ha='right', fontsize=8)
    ax1.set_ylim(0, 1.05)
    ax1.set_ylabel('Normalized ROUGE-L')
    ax1.legend(loc='upper right')
    ax1.set_title(f'Per-field normalized ROUGE-L -- Study# {study_id}: {paper_name[:60]}')
    ax1.axhline(0.5, color='grey', ls='--', lw=0.8, alpha=0.6)

    mnames = list(method_overall_scores.keys())
    omeans = [method_overall_scores[m] for m in mnames]

    # Append MAS step-1 artifact metrics to overall panel if available.
    if 'mas_artifact_scores' in globals() and 'mas_step1_value_recall' in mas_artifact_scores:
        mnames.append('mas_step1_value_recall')
        omeans.append(mas_artifact_scores['mas_step1_value_recall'])
    if 'mas_artifact_scores' in globals() and 'mas_step1_value_rougeL' in mas_artifact_scores:
        mnames.append('mas_step1_value_rougeL')
        omeans.append(mas_artifact_scores['mas_step1_value_rougeL'])

    bars = ax2.bar(mnames, omeans, color=[COLORS.get(m, 'steelblue') for m in mnames], alpha=0.85, width=0.45)
    for bar, name, val in zip(bars, mnames, omeans):
        label = f'{val:.3f}' if pd.notna(val) else 'nan'
        ax2.text(bar.get_x() + bar.get_width() / 2, (val if pd.notna(val) else 0) + 0.02, label, ha='center', fontsize=11, fontweight='bold')
        if name in ('mas_step1_value_recall', 'mas_step1_value_rougeL'):
            bar.set_hatch('//')

    ax2.set_ylim(0, 1.05)
    ax2.set_ylabel('Score')
    ax2.set_title('Overall normalized ROUGE-L + MAS Step-1 artifact diagnostics')
    ax2.axhline(0.5, color='grey', ls='--', lw=0.8, alpha=0.6)

    plt.tight_layout()
    plot_path = get_method_output_path(n, 'evaluation_plot', n_fields).replace('.csv', '.png')
    plt.savefig(plot_path, dpi=150, bbox_inches='tight')
    plt.show()
    print(f'Plot saved: {plot_path}')
````
